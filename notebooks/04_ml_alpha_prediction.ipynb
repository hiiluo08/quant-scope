# 04 ML Alpha Prediction

## Goal
Compare XGBoost and LightGBM regressors on a chronologically split 5-day forward-return prediction task, then inspect the selected champion model's top-5 ranking strategy backtest using Week 4 cost accounting.

## Guardrails
- Labels are future outcomes (`forward_return_5d`), never input features.
- Model selection is validation-only (lowest validation RMSE); test metrics are reported once after champion selection.
- The 5-day embargo prevents target overlap between train/validation and validation/test splits.
- The ranking backtest shifts prediction signals by one trading day and applies 5 bps transaction costs and 5 bps slippage.
- Results are educational research evidence, not investment advice or proof of future alpha.

## 1. Load Stored Model Artifacts and Manifests

```python
from pathlib import Path
import pandas as pd
import json

from ml.training.storage import load_model_manifest, load_predictions

models_root = Path("../data/artifacts/models")
if not models_root.exists():
    models_root = Path("data/artifacts/models")

manifests = []
for path in sorted(models_root.glob("model_id=*/manifest.json")):
    model_id = path.parent.name.removeprefix("model_id=")
    manifest = load_model_manifest(model_id, models_root)
    test = load_predictions(model_id, "test", models_root)
    assert test["date"].min() >= pd.Timestamp(manifest["split_dates"]["test_start"])
    assert test["prediction"].notna().all()
    manifests.append(manifest)

summary_df = pd.DataFrame([
    {
        "model_id": item["model_id"],
        "family": item["family"],
        "val_rmse": item["metrics"]["validation"]["rmse"],
        "val_mae": item["metrics"]["validation"]["mae"],
        "val_rank_ic": item["metrics"]["validation"]["rank_ic"],
        "val_dir_acc": item["metrics"]["validation"]["directional_accuracy"],
        "test_rmse": item["metrics"]["test"]["rmse"],
        "test_mae": item["metrics"]["test"]["mae"],
        "test_rank_ic": item["metrics"]["test"]["rank_ic"],
        "test_dir_acc": item["metrics"]["test"]["directional_accuracy"],
    }
    for item in manifests
])
summary_df
```

## 2. Inspect Champion Model Ranking Backtest

```python
backtests_root = Path("../data/artifacts/backtests")
if not backtests_root.exists():
    backtests_root = Path("data/artifacts/backtests")

ml_backtest_meta = None
for p in sorted(backtests_root.glob("backtest_id=*/metadata.json")):
    meta = json.loads(p.read_text())
    if meta.get("strategy_name") == "ml_top_5_rank_v1":
        ml_backtest_meta = meta
        break

if ml_backtest_meta:
    print("Strategy:", ml_backtest_meta["strategy_name"])
    print("Champion Model ID:", ml_backtest_meta.get("champion_model_id"))
    print("Test Period:", ml_backtest_meta["start_date"], "to", ml_backtest_meta["end_date"])
    print("--- Strategy Metrics ---")
    for k, v in ml_backtest_meta["metrics"].items():
        print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")
    print("--- Benchmark Metrics (SPY) ---")
    for k, v in ml_backtest_meta["benchmarks"]["spy"].items():
        print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")
    print("--- Benchmark Metrics (Equal Weight Buy & Hold) ---")
    for k, v in ml_backtest_meta["benchmarks"]["equal_weight_buy_and_hold"].items():
        print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")
```
